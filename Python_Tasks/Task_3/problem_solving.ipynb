Q1. Find the Frequency of Each Word in a Sentence  
Write a Python program to count how many times each word appears in a given sentence and store the result in a dictionary.   
Example Input:  
python is easy and python is powerful  
Expected Output:  
{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}


##### Pseudocode  
1.Take a sentence as input.    
2.Split the sentence into individual words.   
3.Create an empty dictionary.     
4.Iterate through each word in the sentence.     
5.If the word exists in the dictionary, increase its count by 1.Otherwise, add the word to the dictionary with a count of 1.      
6.Return and print the dictionary.    

```python
def word_frequency(sentence):
    freq = {}
    
    for word in sentence.split():
        if word in freq:
            freq[word]+=1
        else:
            freq[word]=1
    return freq

print(word_frequency("python is easy and python is powerful"))

#Additional Test Case
print(word_frequency("data analytics and data engineering"))
    
```

Q2. Find the Frequency of Each Character in a Word  
Write a Python program to count how many times each character appears in a given word and store the result in a dictionary.  
Example Input:  
banana  
Expected Output:  
{'b': 1, 'a': 3, 'n': 2}


##### Pseudocode  
1.define a function that takes word as input.    
2.Create an empty dictionary.     
3.Iterate through each character in the word.     
4.If the character exists in the dictionary, increase its count by 1.  
5.Otherwise, add the character to the dictionary with a count of 1.      
5.Return and print the dictionary.    

```python
def character_count(word):
    freq = {}
    for ch in word:
        if ch in freq:
            freq[ch]+=1
        else:
            freq[ch]=1
    return freq

print(character_count("banana"))

#Additional Test Case
print(character_count("apple"))    
```

Q3. Find the First Non-Repeating Character in a Word  
Write a Python program to find the first character that appears only once in a given word. If no such character exists, return None.  
Example Input:  
swiss  
Expected Output:  
w


##### Pseudocode   
1.define a funtion that takes word as input.  
2.check each character.  
3.count how many times the current character appears in the word. 
4.If count is 1,return character.  
5.If no character appears only once,return None   

```python
def first_non_repeating(word):
    for ch in word:
        if word.count(ch)==1:
            return ch 
    return None

print(first_non_repeating("swiss"))

#Additional Test case
print(first_non_repeating("aabbb"))
```

Q4. Check Whether Two Strings Are Anagrams  
Write a Python program to check whether two strings are anagrams.  
Two strings are anagrams if they contain the same characters with the same frequencies, regardless of their order.  
Example Input:  
s1 = "listen"  
s2 = "silent"  
Expected Output:  
True

##### Pseudocode    
1.Take two strings as input.  
2.Convert both strings into lowercase and remove space.  
3.sort the characters of both strings.  
4.If they are equal, return True.  
5.Otherwise,return False.  

```python
def anagram(s1,s2):
    s1=s1.replace(" ","").lower()
    s2=s2.replace(" ","").lower()
    if(sorted(s1)==sorted(s2)):
        return True
    else:
        return False

print(anagram("listen","silent"))  

#Additional Test Case:
print(anagram("secure","rescue"))
```

Q5. Find the First Repeating Character in a String  
Write a Python program to find the first character whose second occurrence is encountered while reading the string from left to right.  
If no character repeats, return None.  
Example Input:  
programming 
Expected Output:  
r


##### Pseudocode  
1.Define a function that takes a string.  
2.Create an empty set.  
3.Check each character.
4.If the character is already in the set, return it.   
5.Otherwise, add the character to the set.   
6.If no character repeats, return None.  

```python
def first_repeating(word):
    word = word.replace(" ","").lower()
    first_char = set()
    
    for ch in word:
        if ch in first_char:
            return ch
        first_char.add(ch)
    return None

print(first_repeating("programming")) 

#Additional Test Case
print(first_repeating("Advanced"))
```

Q6. Move All Zeros to the End of a List  
Write a Python program to move all zeros to the end of a list while maintaining the original order of non-zero elements.  
Example Input:  
[0, 3, 0, 1, 2, 0, 4]  
Expected Output:  
[3, 1, 2, 4, 0, 0, 0]  


##### Pseudocode  
1.Define a function that takes a list.  
2.Create an empty list for non-zero values.  
3.Count the zeros in the original list.  
4.Add the zeros to the end of the new list.  
5.Return the new list and print the result.  

```python
def list_of_numbers(nums):
    res = []
    
    for num in nums:
        if num!=0:
            res.append(num)
    res.extend([0]*nums.count(0))
    return res

print(list_of_numbers([0, 3, 0, 1, 2, 0, 4] ))

#Additional Test Case
print(list_of_numbers([0, 0, 0, 1, 2, 0, 4,8,2,0] ))
```

Q7. Find the Second Highest Distinct Number in a List  
Write a Python program to find the second-largest distinct number in a given list.  
Duplicate values should not be considered separate ranks.  
If a second distinct number does not exist, return None.  
Example Input:  
[10, 20, 40, 30, 40, 20]  
Expected Output:  
30


##### Pseudocode  
1.Create an empty set for storing Unique numbers.  
2.If the set has less than 2 numbers then return None,else sort in ascending order and return the last 2nd element to get the second largest number.  
3.print the result.

```python
def second_largest_distinct_num(nums):
    unique_nums = set(nums)
    if(len(unique_nums)<2):
        return None
    return sorted(unique_nums)[-2]

print(second_largest_distinct_num([10, 20, 40, 30, 40, 20]))

#Additional Test Case
print(second_largest_distinct_num([100, 200, 400, 300, 400, 200]))
```

Q8. Remove Duplicates Without Changing the Original Order   
Write a Python program to remove duplicate elements from a list while   preserving the order of their first occurrence.  
Example Input:  
[4, 2, 4, 5, 2, 7, 5]  
Expected Output:  
[4, 2, 5, 7]  


##### Pseudocode  
1.create an empty result list.  
2.Check each number in the original list.    
3.Add the number if it is not already in the result list.  
4.Return the result list.

```python
def remove_duplicates(nums):
    result=[]
    
    for num in nums:
        if num not in result:
            result.append(num)
    return result

print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

#Additional Test Case
print(remove_duplicates([1,4, 1, 4, 5, 2, 6,2, 5]))
```

Q9. Find the Missing Number from 1 to N  
Write a Python program to find the missing number in a list containing numbers from 1 to N.  
The list contains exactly one missing number, and all remaining numbers appear only once.  
Example Input:  
nums = [1, 2, 4, 5, 6]   
N = 6   
Expected Output:  
3  


##### Pseudocode  
1.Define a function that takes a list and N.  
2.for finding sum of consecutive numbers n*(n+1)/2 is used.  
3.calculate the sum of nums in list using sum().   
4.expected sum - actual sum gives the missing number.  
5.return the missing number.  

```python
def missing_number(nums,n):
    expected_sum = n*(n+1)/2
    actual_sum = sum(nums)
    
    return int(expected_sum-actual_sum)

print(missing_number([1, 2, 4, 5, 6], 6))

#Additional Test Case
print(missing_number([1, 2,3, 4, 5, 6,8,9], 9))

```

Q10. Find Two Numbers Whose Sum Equals a Target  
Write a Python program to find the first pair of numbers whose sum equals a given target.  
Return the pair as a list. If no valid pair exists, return an empty list.  
Return the pair whose second element is encountered earliest when scanning the input from left to right. If multiple earlier elements qualify, choose the earliest one.  
Example Input:  
nums = [2, 7, 11, 15]  
target = 9  
Expected Output:  
[2, 7]  


##### Pseudocode  
1.Define a function that takes a list and target.  
2.Check each number from left to right.  
3.For each number, check the earlier numbers in order.  
4.If their sum equals the target, return the pair.  
5.If no pair matches, return an empty list.  

```python
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i):
            if nums[j] + nums[i] == target:
                return [nums[j], nums[i]]
    return []

print(two_sum([2, 7, 11, 15], 9))

#Additional Test Case
print(two_sum([1, 4, 6], 20))
```

Q11. Find the Number Closest to Zero  
Write a Python program to find the number closest to zero in a given list.  
If two numbers are equally close to zero, return the positive number.  
Example Input:  
[-8, -3, 2, 5, -2]  
Expected Output:   
2


##### Pseudocode  
1.Find the number with the smallest absolute value.  
2.If two numbers are equally close to zero, choose the positive number.  
3.Return the selected number.  

```python
def closest_to_zero(nums):
    return min(nums, key=lambda num: (abs(num), -num))

print(closest_to_zero([-8, -3, 2, 5, -2]))

#Additional Test Case
print(closest_to_zero([-4, -1, 1, 6]))
```

Q12. Find Common Elements Between Two Lists    
Write a Python program to find the common elements between two lists.   
Each common element should appear only once in the result, maintaining the order of its first occurrence in the first list.   
Example Input:  
list1 = [1, 2, 3, 4, 5]  
list2 = [3, 4, 5, 6, 7]   
Expected Output:   
[3, 4, 5]


##### Pseudocode    
1.Define a function that takes two lists.  
2.Create an empty result list.  
3.Check each element in the first list.  
4.If the element exists in the second list and is not already in the result, add it.  
5.Return the result list.

```python
def common_elements(list1,list2):
    result = []
    for num in list1:
        if num in list2 and num not in result:
            result.append(num)
    return result

print(common_elements([1, 2, 3, 4, 5],[3, 4, 5, 6, 7] ))

#Additional Test Case
print(common_elements([2,4,5,6],[2,4,7,8]))
```

Q13. Rotate a List to the Right by K Positions  
Write a Python program to rotate the elements of a list to the right by K positions.  
Elements shifted beyond the last position should move to the beginning of the list.  
Example Input:  
nums = [1, 2, 3, 4, 5]  
K = 2  
Expected Output:  
[4, 5, 1, 2, 3]


##### Pseudocode  
1.If the list is empty, return an empty list.  
2.Calculate K modulo the list length.  
3.If K is zero, return a copy of the original list.  
4.Otherwise, place the last K elements at the beginning.  
5.return the list

```python
def rotate(nums, k):
    if not nums:
        return []
    k = k % len(nums)
    if k == 0:
        return nums[:]

    return nums[-k:] + nums[:-k]

print(rotate([1, 2, 3, 4, 5], 2))

#Additional Test Case
print(rotate([11, 21, 34], 1))
```

Q14. Find the Maximum Sum of Two Consecutive Elements  
Write a Python program to find the maximum sum of any two consecutive elements in a list.  
Assume the list contains at least two elements.  
Example Input:  
[4, 7, 2, 9, 5]  
Expected Output:  
14


##### Pseudocode  
1.Add the first two numbers and store the sum as the maximum.  
2.Check each pair of consecutive numbers in the list.  
3.If the sum of a pair is greater than the current maximum, update the maximum.  
4.Return the maximum sum.

```python
def max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]

    for i in range(1, len(nums) - 1):
        current_sum = nums[i] + nums[i + 1]
        if current_sum > max_sum:
            max_sum = current_sum

    return max_sum

print(max_consecutive_sum([4, 7, 2, 9, 5]))

#Additional Test Case
print(max_consecutive_sum([-5, -2, -8]))
```

Q15. Find All Pairs of Numbers Whose Sum Equals a Target
Write a Python program to find all unique pairs of numbers whose sum equals a given target.  
Each pair should contain two distinct list positions, and duplicate pairs should not be included.  
Return each pair in ascending numerical order. Return the list of pairs sorted by their first element. If no valid pairs exist, return an empty list.
Example Input:  
nums = [2, 4, 3, 5, 7, 8, 1]  
target = 9  
Expected Output:  
[[1, 8], [2, 7], [4, 5]]  


##### Pseudocode  
1.Create an empty result list.  
2.Check every pair of elements at different positions.  
3.If their sum equals the target, arrange the pair in ascending order.  
4.Add the pair only if it is not already in the result list.  
5.Sort the result by the first element of each pair.  
6.Return the result list.

```python
def all_pairs(nums, target):
    result = []

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = sorted([nums[i], nums[j]])
                if pair not in result:
                    result.append(pair)

    result.sort(key=lambda pair: pair[0])
    return result

print(all_pairs([2, 4, 3, 5, 7, 8, 1], 9))

#Additional Test Case
print(all_pairs([1,1,2, 2,3, 3], 5))
```
