### Q1. Find the Shortest Word in a Sentence
Write a Python program to find the shortest word in a given sentence.
Example input:  
Python makes coding easy  
Expected output:  
easy  


```python
# 1. Observe:
# Input: "Python makes coding easy"
# Expected Output: easy
# We need to find the shortest word in the given sentence.

# 2. Understand:
# First, we split the sentence into individual words.
# Then, we compare the length of each word and identify the shortest one.

# 3. Breakdown:
# Step 1: Store the sentence in a variable.
# Step 2. Split the sentence into words.
# Step 3. Assume the first word is the shortest.
# Step 4. Check each word in the list.
# Step 5. If a word is shorter, update the shortest word.
# Step 6. Print the shortest word.

# 4. Logic:
# Start
# Take a sentence
# Split the sentence into words
# Set shortest = first word
# For each word in the list:
#     If length of word < length of shortest:
#         Set shortest = word
# Print shortest
# Stop

# 5. Code:
sentence = "Python makes coding easy"
words = sentence.split()
shortest = words[0]

for word in words:
    if len(word) < len(shortest):
        shortest = word

print(shortest)

# 6. Test:
# Test Case 1:
# Input: "Python makes coding easy" 
# Expected Output: easy

# Test Case 2:
# Input: "I love programming"
# Expected Output: I
```

### Q7. Count Consonants in a String
Write a Python program to count the consonants in a string. Ignore spaces, digits, and special characters.  
Example input:  
Python 123!  
Expected output:  
5


```python
# 1. Observe:
# Input: "Python 123!"
# Expected Output: 5
# We need to count the consonants in a string.
# Spaces, digits, and special characters should be ignored.

# 2. Understand:
# Consonants are alphabetic characters that are not vowels.
# We check each character to see whether it is a letter.
# If it is a letter and not a vowel/digit/special character, we increase the count by 1.

# 3. Breakdown:
# Step 1: Store the string in a variable.
# Step 2: Initialize count to 0.
# Step 3: Check each character in the string.
# Step 4: If the character is a letter and not a vowel/digit/special character, increase count by 1.
# Step 5: Print the total number of consonants.

# 3. Logic:
# Start
# Take a string
# Set count = 0
# For each character in the string:
#     If character is a letter and not a vowel:
#         Increase count by 1
# Print count
# Stop

# 5. Code:
text = "Python 123!"
count = 0

for char in text:
    if char.isalpha() and char.lower() not in "aeiou":
        count = count + 1

print(count)

# 6. Test:
# Test Case 1:
# Input: "Python 123!"
# Expected Output: 5

# Test Case 2:
# Input: "Hello World!"
# Expected Output: 7
```

### Q8. Find the Most Frequent Character
Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.  
Example input:  
banana  
Expected output:  
a


```python
# 1. Observe:
# Input: "banana"
# Expected Output: a
# We need to find the character that appears most frequently in a string.
# Spaces should be ignored.
# If multiple characters have the same highest frequency, return the first one encountered.

# 2. Understand:
# We need to count how many times each character appears in the string.
# We can use a dictionary to store each character and its frequency.
# Then, we find the character with the highest frequency.
# If there is a tie, we keep the character that appears first.

# 3. Breakdown:
# Step 1: Store the string in a variable.
# Step 2: Remove spaces from the string.
# Step 3: Create an empty dictionary to store character frequencies.
# Step 4: Count the occurrences of each character.
# Step 5: Initialize the maximum count to 0.
# Step 6: Find the character with the highest frequency.
# Step 7: Print the most frequent character.

# 4. Logic:
# Start
# Take a string
# Remove spaces
# Create an empty dictionary
# For each character in the string:
#     Update its frequency in the dictionary
# Set max_count = 0
# Set most_frequent_char = empty string
# For each character and its count in the dictionary:
#     If count > max_count:
#         Set max_count = count
#         Set most_frequent_char = character
# Print most_frequent_char
# Stop

# 5. Code:
text = "banana"
text = text.replace(" ", "")
freq = {}

for char in text:
    freq[char] = freq.get(char, 0) + 1

max_count = 0
most_frequent_char = ""

for char, count in freq.items():
    if count > max_count:
        max_count = count
        most_frequent_char = char

print(most_frequent_char)

# Step 6: Test
# Test Case 1:
# Input: "banana"
# Expected Output: a

# Test Case 2:
# Input: "aabbc"
# Expected Output: a
```

### Q10. Find All Characters Appearing More Than Once
Write a Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces. 
Example input:   
programming  
Expected output:  
['r', 'g', 'm']


```python
# 1. Observe:
# Input: "programming"
# Expected Output: ['r', 'g', 'm']

# 2. Understand:
# First, we count how many times each character appears in the string.
# Then, we check which characters have a frequency greater than 1.
# We add each repeated character to a list only once.
# This helps us maintain the order in which the characters first appear.

# 3. Breakdown:
# Step 1: Store the string in a variable.
# Step 2: Remove spaces from the string.
# Step 3: Create an empty dictionary to store character frequencies.
# Step 4: Count the occurrences of each character.
# Step 5: Create an empty list to store repeated characters.
# Step 6: Check each character in the original order.
# Step 7: If a character appears more than once and is not already in the list,
#    add it to the list.
# Step 8: Print the list of repeated characters.

# 4. Logic:
# Start
# Take a string
# Remove spaces
# Create an empty dictionary
# For each character in the string:
#     Update its frequency in the dictionary
# Create an empty list called repeated
# For each character in the string:
#     If frequency of character > 1 and character is not in repeated:
#         Add character to repeated
# Print repeated
# Stop

# 5. Code:
text = "programming"
text = text.replace(" ", "")
freq = {}

for char in text:
    freq[char] = freq.get(char, 0) + 1

repeated = []

for char in text:
    if freq[char] > 1 and char not in repeated:
        repeated.append(char)

print(repeated)

# 6. Test:
# Test Case 1:
# Input: "programming"
# Expected Output: ['r', 'g', 'm']

# Test Case 2:
# Input: "aabbcc"
# Expected Output: ['a', 'b', 'c']
```

### Q15. Create a Dictionary of Numbers and Their Even/Odd Status
Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.  
Example input:  
[1, 2, 3, 4, 5]  
Expected output:  
{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'} 


```python
# 1. Observe:
# Input: [1, 2, 3, 4, 5]
# Expected Output: {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}
# We need to create a dictionary where each number is a key and its even or odd status is the value.

# 2. Understand:
# We need to check whether each number is even or odd.
# If a number is divisible by 2, it is even.
# Otherwise, it is odd.
# We store each number and its status in a dictionary.

# 3. Breakdown:
# Step 1: Store the numbers in a list.
# Step 2: Create an empty dictionary.
# Step 3: Check each number in the list.
# Step 4: If the number is divisible by 2, assign "Even".
# Step 5: Otherwise, assign "Odd".
# Step 6: Store the number and its status in the dictionary.
# Step 7: Print the dictionary.

# 4. Logic:
# Start
# Take a list of numbers
# Create an empty dictionary
# For each number in the list:
#     If number % 2 == 0:
#         Set status = "Even"
#     Otherwise:
#         Set status = "Odd"
#     Store number as key and status as value
# Print dictionary
# Stop

# 5. Code:
numbers = [1, 2, 3, 4, 5]
result = {}

for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"

print(result)

# 6. Test:
# Test Case 1:
# Input: [1, 2, 3, 4, 5]
# Expected Output: {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}

# Test Case 2:
# Input: [6, 7, 8]
# Expected Output: {6: 'Even', 7: 'Odd', 8: 'Even'}
```

### Q16. Find the Most Frequent Word in a Sentence
Write a Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same word. If multiple words share the highest frequency, return the one that appears first.  
Example input:  
Python is easy and Python is powerful  
Expected output:  
python


```python
# 1. Observe:
# Input: "Python is easy and Python is powerful"
# Expected Output: python
# We need to find the word that appears most frequently in a sentence.
# Uppercase and lowercase versions of a word should be treated as the same.
# If multiple words have the same highest frequency, return the one that appears first.

# 2. Understand:
# First, we convert the sentence to lowercase so that words like
# "Python" and "python" are treated as the same word.
# Then, we count how many times each word appears.
# Finally, we find the word with the highest frequency.
# If there is a tie, we keep the word that appears first.

# 3. Breakdown:
# Step 1: Store the sentence in a variable.
# Step 2: Convert the sentence to lowercase.
# Step 3: Split the sentence into words.
# Step 4: Create an empty dictionary to store word frequencies.
# Step 5: Count the occurrences of each word.
# Step 6: Find the word with the highest frequency.
# Step 7: Print the most frequent word.

# 4. Logic:
# Start
# Take a sentence
# Convert the sentence to lowercase
# Split the sentence into words
# Create an empty dictionary
# For each word in the list:
#     Update its frequency in the dictionary
# Set max_count = 0
# Set most_frequent_word = empty string
# For each word and its count in the dictionary:
#     If count > max_count:
#         Set max_count = count
#         Set most_frequent_word = word
# Print most_frequent_word
# Stop

# 5. Code:
sentence = "Python is easy and Python is powerful"
sentence = sentence.lower()
words = sentence.split()
freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

max_count = 0
most_frequent_word = ""

for word, count in freq.items():
    if count > max_count:
        max_count = count
        most_frequent_word = word

print(most_frequent_word)

# 6. Test:
# Test Case 1:
# Input: "Python is easy and Python is powerful"
# Expected Output: python

# Test Case 2:
# Input: "Data science data analysis"
# Expected Output: data
```

```python
# 1.Observe:
# Input: "data science uses data and science tools"
# Expected Output: ['data', 'science']
# We need to find all words that appear more than once in a sentence.
# Each repeated word should be displayed only once.
# Uppercase and lowercase versions of a word should be treated as the same.

# 2. Understand:
# First, we convert the sentence to lowercase so that words like "Data" and "data" are treated as the same word.
# Then, we count how many times each word appears.
# Finally, we identify the repeated words and maintain their first-appearance order.

# 3. Breakdown:
# Step 1: Store the sentence in a variable.
# Step 2: Convert the sentence to lowercase.
# Step 3: Split the sentence into words.
# Step 4: Create an empty dictionary to store word frequencies.
# Step 5: Count the occurrences of each word.
# Step 6: Create an empty list to store repeated words.
# Step 7: Check each word in the original order.
# Step 8: If a word appears more than once and is not already in the list,
#         add it to the list.
# Step 9: Print the list of repeated words.

# 4. Logic:
# Start
# Take a sentence
# Convert the sentence to lowercase
# Split the sentence into words
# Create an empty dictionary
# For each word in the list:
#     Update its frequency in the dictionary
# Create an empty list called repeated
# For each word in the list:
#     If frequency of word > 1 and word is not in repeated:
#         Add word to repeated
# Print repeated
# Stop

# 5. Code:
sentence = "data science uses data and science tools"
sentence = sentence.lower()
words = sentence.split()
freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

repeated = []

for word in words:
    if freq[word] > 1 and word not in repeated:
        repeated.append(word)

print(repeated)

# 6. Test:
# Test Case 1:
# Input: "data science uses data and science tools"
# Expected Output: ['data', 'science']

# Test Case 2:Replace the sentence above with:
# Input: "Apple banana apple ORANGE banana"
# Expected Output: ['apple', 'banana']
```

### Q20. Find the Longest Word That Appears Only Once
Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.  
Example input:  
Python makes coding useful Python  
Expected output:  
coding  


```python
# 1. Observe:
# Input: "Python makes coding useful Python"
# Expected output: coding

# 2. Understand:
# Convert the sentence to lowercase so uppercase and lowercase
# versions of a word are treated as the same word.
# Count the frequency of each word using a dictionary.
# Then find the longest word whose frequency is exactly 1.
# If multiple words have the same maximum length, keep the first one.

# 3. Breakdown:
# Step 1: Take the input sentence.
# Step 2: Convert the sentence to lowercase.
# Step 3: Split the sentence into words.
# Step 4: Count each word using a dictionary.
# Step 5: Initialize the longest word as an empty string.
# Step 6: Check each word whose frequency is 1.
# Step 7: If its length is greater than the current longest word,
#         update the longest word.
# Step 8: Print the longest word.

# 4. Logic:
# Start
# Take the input sentence
# Convert the sentence to lowercase
# Split the sentence into words
# Create an empty dictionary called freq
# For each word in words:
#     Update its count in freq
# Set longest = ""
# For each word in words:
#     If freq[word] equals 1:
#         If length of word > length of longest:
#             Set longest = word
# Print longest
# Stop

# 5. Code:
sentence = "Python makes coding useful Python"

sentence = sentence.lower()
words = sentence.split()

freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

longest = ""

for word in words:
    if freq[word] == 1:
        if len(word) > len(longest):
            longest = word

print(longest)

# 6. Test:
# Test case 1:
# Input: "Python makes coding useful Python"
# Output: coding

# Test case 2:Replace the sentence above with:
# sentence = "cat tiger lion apple"
# Expected output: tiger
```
