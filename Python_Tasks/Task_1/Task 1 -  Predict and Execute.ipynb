#### Q20

```python
a = [10, 20]
b = [10, 20]
print(a == b)
print(a is b)

# Explanation:
# Both lists contain the same elements in the same order,so a == b returns True because == compares the values.
# However, a and b are two separate list objects stored in memory.
# Therefore, a is b returns False because is checks object identity.

# Output:
# True
# False
```

#### Q21

```python
a = [10, 20]
b = a
print(a == b)
print(a is b)

# Explanation:
# The variable b is assigned to a, so both variables refer to the same list object in memory.
# Therefore, a == b returns True because their values are equal.
# Also, a is b returns True because both variables refer to the exact same object.

# Output:
# True
# True
```

#### Q35

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

# Explanation:
# Initially, a contains two inner lists: [1, 2] and [3, 4].
# copy.copy(a) creates a shallow copy, so b is a new outer list, but both a and b share references to the same inner lists.
# The statement b[0][1] = 99 changes the second element of the first inner list. 
# Since that inner list is shared, the change is also visible through a.
# Therefore, print(a) displays [[1, 99], [3, 4]].

# Output:
# [[1, 99], [3, 4]]
```

#### Q36

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

# Explanation:
# Initially, a contains two inner lists: [1, 2] and [3, 4].
# copy.deepcopy(a) creates a completely independent copy, including the inner lists.
# When b[0][1] = 99 is executed, only the copied list b is modified.
# The original list a remains unchanged because its inner lists are separate objects from those in b.
# Therefore, print(a) displays [[1, 2], [3, 4]].

# Output:
# [[1, 2], [3, 4]]
```
